```python
import pandas as pd
import numpy as np
import os
```

```python
'''
3. Programación en Python | Hola Mundo
'''

print('Hola Mundo')
print('Mi nombre es Eduardo')

# Para dar un salto de línea ocupamos '\n'

print('Hola Mundo \nMi nombre es Eduardo')
```

```python
'''
4. Programación en Python | Asignación de Valores
'''
# Para imprimir las variables, se utiliza print sin comillas
# Para imprimir el tipo de variable se ocupa 'type' dentro del print

numero1 = 10 # Variable de tipo entero
print(numero1)
print(type(numero1))

numero2 = 10.56 # Variable de tipo flotante
print(numero2)
print(type(numero2))

# Para declarar variables de cadenas (strings) se utilizan las comillas (simples o dobles)

cadena1 = 'a' # Variable de tipo cadena
print(cadena1)
print(type(cadena1))


cadena2 = 'Estoy "estudiando"' # Variable de tipo cadena
print(cadena2)
print(type(cadena2))

cadena3 = True # Variable de tipo booleano
print(cadena3)
print(type(cadena3))

# Asignar valor a variables y hacer operaciones con las variables

num1 = 10
num2 = 6.7

suma = num1 + num2
print(f'La suma de {num1} + {num2} es igual a {suma}')

# Podemos ocupar números con variables sin problemas
# Python respeta la jerarquía de operaciones de las matemáticas (PEMDAS)

operacion = num1 + num2 *10/6
print(f'La operación de sumar {num1} + {num2} multiplicado por 10 y dividido entre 6 es igual a {operacion}')

# Podemos concatenar el print de los resultados
print('El resultado de la operacion es:', operacion)

# Python tiene tipado dinámico, es decir, si asignas un valor nuevo a la misma variable Python toma la última asignación

valor = 10 # Declaramos la variable 'valor' por primera vez
print(valor)

valor = 'Eduardo' # Al declarar la variable 'valor' una segunda vez, la primera variable es reemplazada
print(valor)
```

```python
''' 
6. programación en Python | Operadores Aritméticos
'''
# Se pueden asignar directamente los valores

operacion1 = 10 + 5
print(operacion1)

# Se pueden utilizar variables con valores almacenados

a = 10
b = 3

operacion2 = a - b
print(operacion2)

# División flotante en Python se realiza con un único '/'

operacion3 = a/b
print(operacion3)

# División entera en Python se realiza con dos '//'

operacion4 = a//b
print(operacion4)

# La exponenciación se realiza con dos '**'

operacion5 = a**b
print(operacion5)

# El módulo nos regresa el residuo de una división y se realiza con '%'

operacion6 = a%b
print(operacion6)

# Prioridad de los operadores aritméticos (PEMDAS)
# 1. Parentésis ()
# 2. Exponenciación **
# 3. Multiplicación, División y Módulo *, /, %
# 4. Suma y Resta, +,- 
```

```python
'''
7. Programación en Python | Operadores Relacionales
'''
# Se utilizan para establecer una relación entre dos valores
# Compara estos valores entre sí y produce un resultado de verdadero o falso
# Tienen el mismo nivel en su evaluación
# Los operadores relacionales tienen menor prioridad que los aritméticos

# Los operadores relacionales son:
# >,< Mayor o menor que
# >=, <=, Mayor o igual que y Menor o igual que
# != Diferente que
# == Igual que

resultado1 = 10 > 20 # ¿Es 10 mayor que 20?
print(resultado1)

resultado2 = 10 < 20 # ¿Es 20 mayor que 10?
print(resultado2)

resultado3 = 10 == 20 # ¿Es 10 igual que 20?
print(resultado3)

resultado4 = 10 != 20 # ¿Es 10 diferente que 20?
print(resultado4)

resultado5 = 17 <= 17 # ¿Es 17 menor o igual que 17?
print(resultado5)

resultado6 = 17 >= 17 # ¿Es 17 mayor o igual que 17?
print(resultado6)

# Podemos combinar operadores aritméticos con operadores relacionales

a = 10
b = 20
c = 30

resultado7 = a+b == c # ¿Es 10 + 20 igual que 30?
print(resultado7)
```

```python
'''
8. Programación en Python | Operadores Lógicos
'''

# Permiten construir expresiones lógicas, se obtiene como resultado un booleano
# Los operadores lógicos son:
# 'and' - Conjunción
# 'or' - Disyunción
# 'not' - Negación

a = 10
b = 15
c = 20

operador1 = (a<b) and (b<c) # ¿Es 15 mayor que 10 y 20 mayor que 15?
print(operador1)

operador2 = (a>b) and (b<c) # ¿Es 10 mayor que 15 y 20 mayor que 15?
print(operador2)

operador3 = (a>b) or (b<c) # ¿Es 10 mayor que 15 o 20 mayor que 15?
print(operador3)

operador4 = not((a>b) or (b<c)) # Negación de -> ¿Es 10 mayor que 15 o 20 mayor que 15?
print(operador4)
```

```python
'''
9. Programación en Python | Operadores de Asignación
'''

# Nos ayudan a acortar el código y a que se vea más estético

a = 0 # Primero se debe de dar un valor inicial a la variable

a += 5 # Esto es una suma en asignación
a -= 2 # Esto es una resta en asignación
a *= 3 # Esto es una multiplicación en asignación
a /= 3 # Esto es una división en asignación
a **= 2 # Esto es una exponenciación en asignación
a %= 2 # Esto es un modulo en asignación

print(f'Partimos de un a = 0 -> +5-2*3/3**2%2 es igual a {a}')
```

```python
'''
11. Programación en Python | Entrada de Datos
'''

nombre = input('Digite su nombre: ')
print(f'Hola {nombre}')

numero1 = int(input('Digite un número: '))
print(f'Su número + 7 es igual a {numero1 + 7} en tipo entero (int)')

numero1 = float(input('Digite un número: '))
print(f'Su número + 7 es igual a {numero1 + 7} en tipo flotante')
```

```python
'''
12. Programación en Python | Funciones Integradas
'''
# Sirven para hacer conversiones entre los tipos de datos
# Podemos convertir de una cadena a un valor numérico
n = int('10') # Convertimos a entero una cadena de texto
print(n+2)
m = float('15.7') # Convertimos a flotante una cadena de texto
print(m+7)

# Podemos cambiar de base numérica
k = bin(10)
print(f'Transformamos el 10 en binario -> {k}') # Python representa el binario con un '0b' antes del número
l = hex(30)
print(f'Transformamos el 30 en hexadecimal -> {l}') # Python representa el hexadecimal con un '0x' antes del número

# Podemos convertir distintas bases a enteros
f = int('0b1010', 2) # Convertimos de binario a entero
print(f)
j = int('0x14', 16) # Convertimos de hexadecimal a entero
print(j)

# Podemos obtener el absoluto y redondear hacía el más cercano
w = abs(-7) # Valor absoluto
print(w)
z = round(5.9) # Valor redondeado
print(z)

# También podemos medir cuántos caracteres
h = len('Eduardo')
print(h)
```

```python
'''
15. Programación en Python | Ejercicio 3
'''
# Intercambiar el valor de dos variables

m = float(input('Digite el valor de m: ')) # Capturamos la primera variable
n = float(input('Digite el valor de n: ')) # Capturamos la segunda variable

m,n = n,m # Asignamos los valores en la otra variable
print(f'El nuevo valor de m es: {m}')
print(f'El nuevo valor de n es: {n}')
```

```python
'''
16. Programación en Python | Ejercicio 4
'''
# El usuario ingresará el radio de un círculo y el programa le devolverá área y longitud
import math # La librería 'math' tiene el valor de math.pi (3.141569...) 
radio = float(input('Ingrese el radio de la circunferencia: '))

area = math.pi * radio**2 # Calculamos el área del círculo
longitud = 2 * radio * math.pi # Calculamos el perímetro del círculo

print(f'La longitud de la circunferencia es {longitud:.2f}')
print(f'El área de la circunferencia es {area:.2f}')
```

```python
'''
17. Programación en Python | Ejercicio 5
'''
precio = float(input('Ingrese el precio del producto: ')) # Capturamos el precio original del producto
porc_desc = float(input('Ingrese el % de descuento que se va a aplicar: ')) # Capturamos el descuento a aplicar
descuento = precio * (porc_desc/100) # Calculamos el descuento sobre el precio original

total = precio - descuento # Se calcula el nuevo precio con descuento
print(f'Usted deberá de pagar: ${total:.2f}')
```
