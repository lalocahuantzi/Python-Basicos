```python
'''
18. programación en Python | Estructura if-else-if
'''

# Sentencias que sirven para comparar 2 valores
# Si se cumplen ciertas condiciones se realiza o no ciertas operaciones

numero = int(input('Digite un número: '))

if numero > 0: # Evaluamos si el número es mayor que 0, sino se cumple la condición el 'if' NO se ejecuta
    print(f'El número {numero} es positivo')

elif numero == 0: # Evaluamos si el número es igual que 0, sino se cumple la condición el 'elif' NO se ejecuta
    print(f'El número {numero} es cero')

else: # Evaluamos el único caso restante, si el número es menor o igual que 0 se ejecuta el 'else'
    print(f'El número {numero} es negativo')
```

```python
'''
19. Programación en Python | Condicionales Combinados
'''

edad = int(input('Ingrese su edad: '))

if edad > 0 and edad < 100: # Evaluamos si el valor de la edad tiene sentido dentro de cierto rango definido
    print(f'La edad es correcta -> {edad} años')
    if edad >= 18: # Podemos anidar un 'if' dentro de otro 'if', NO hay un número límite definido
        print('Usted es mayor de edad')
    else:
        print('Usted es menor de edad')
else:
    print('La edad es incorrecta')
```

```python
'''
20. Programación en Python | Ejercicio 1
'''
# Ejercicio 1
# Hacer un programa que pida dos número e identifique cuál de ellos es par

num1 = int(input('Digite un número: '))
num2 = int(input('Digite otro número: '))

if num1%2 == 0 and num2%2 == 0: # Nos apoyamos de la función módulo para evaluar la paridad de los dos números
    print('Ambos números son pares')
elif num1%2 == 0 and num2%2 != 0: # Nos apoyamos de la función módulo para evaluar la paridad de los dos números
    print(f'El numero {num1} es par')
elif num1%2 != 0 and num2%2 == 0: # Nos apoyamos de la función módulo para evaluar la paridad de los dos números
    print(f'El numero {num2} es par')
else: # Si la función módulo siempre nos da diferente de 0, signififa que ninguno de los dos números es par y se ejecuta el 'else'
    print('Ninguno de los dos números es par')
```

```python
'''
21. Programación en Python | Ejercicio 2
'''
# Hacer un programa que pida 3 números y diga cuál de los 3 es el mayor

a = float(input('Digite un número: '))
b = float(input('Digite otro número: '))
c = float(input('Digite un último número: '))

if a >= b and a >= c: # Si utilizamos un 'and' significa que AMBAS condiciones se deben de cumplir
    print(f'El número {a} es el mayor')
elif b >= a and b >= c: # Si utilizamos un 'and' significa que AMBAS condiciones se deben de cumplir
    print(f'El número {b} es el mayor')
elif c >= a and c >= b: # Si utilizamos un 'and' significa que AMBAS condiciones se deben de cumplir
    print(f'El número {c} es el mayor')
```

```python
'''
22. Programación en Python | Ejercicio 3
'''
# Hacer un programa que pida un carácter y diga si es vocal o no

letra = input('Digite un carácter: ').lower() # Con .lower() convertimos cualquier cadena de texto a minúsculas

# Si utilizamos un 'or' significa que ALGUNA de las condiciones se deben de cumplir
if letra == 'a' or letra == 'e' or letra == 'i' or letra == 'o' or letra == 'u':
    print('Es una vocal')
else:
    print('No es una vocal')
```

```python
'''
23. Programación en Python | Ejercicio 4
'''
# Hacer un programa que simule el funcionamiento de una calculadora
# El usuario debe de especificar la operación a realizar entre 2 números
# S,s = Suma, R,r = Resta, M,m = Multiplicación, y D,d = División

num1 = float(input('Digite el primer número: '))
num2 = float(input('Digite el segundo número: '))

operacion = input('¿Qué operación quiere realizar?: ').upper() # Con .upper() convertimos cualquier cadena de texto a mayúsculas

if operacion == 'S':
    suma = num1 + num2
    print(f'El resultado de la suma es {suma}')
elif operacion == 'R':
    resta = num1 - num2
    print(f'El resultado de la resta es {resta}')
elif operacion == 'M':
    multiplicacion = num1 * num2
    print(f'El resultado de la multiplicación es {multiplicacion}')
elif operacion == 'D':
    division = num1 / num2
    print(f'El resultado de la división es {division}')
else: # Si se ejecuta este 'else' significa que el usuario ingresó un caracter diferente de S,s,R,r,M,m,D,d
    print('Esa operación no es válida')
```

```python
'''
24. Programación en Python | Ejercicio 5
'''
# Vamos a realizar un programa que simula un cajero automático con sus respectivas opciones
# El saldo inicial del usuario será de $1000
# La lista de opciones del menú será:
# 1.Ingresar dinero en la cuenta, 2.Retirar dinero de la cuenta, 3.Mostrar dinero disponible, 4.Salir

saldo = 1000 # Saldo inicial de la cuenta
print('\t .:Menú:.', flush=True) # 'flush=true' Nos ayuda a imprimir lo que tengamos en ese momento antes de seguir procesando 
print('1. Ingresar dinero de la cuenta')
print('2. Retirar dinero de la cuenta')
print('3. Mostrar disponible')
print('4. Salir')

opcion = int(input('Digite una opción (1,2,3,4): '))
if opcion == 1:
    extra = float(input('¿Cuánto dinero desea ingresar a la cuenta?'))
    saldo += extra # Hacemos una suma en asignación para aumentar el saldo tras un depósito
    print(f'El saldo disponible en la cuenta es de ${saldo}')
elif opcion == 2:
    retiro = float(input('¿Cuánto dinero desea retirar de la cuenta?'))
    if retiro > 1000:
        print('No tiene ese saldo disponible')
    else:
        saldo -= retiro # Hacemos una suma en asignación para reducir el saldo tras un retiro
        print(f'El saldo disponible en la cuenta es de ${saldo}')
elif opcion == 3:
    print(f'El saldo disponible en la cuenta es de ${saldo}')
elif opcion == 4:
    print('Gracias por ocupar este cajero automático')
else:
    print('Opción inválida')
```
