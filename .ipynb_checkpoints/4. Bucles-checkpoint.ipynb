# Bucles

Los bucles son un trozo de código que se va a repetir un número determinado o indeterminado de veces.
<br>
Son ciclos que se van a repetir (loops). En Python existen 2 tipos de bucles *while* y *for*

```python
'''
37. Programación en Python | Bucle While
'''
# Es un bucle con un número indeterminado de iteraciones
# Necesitamos que se siga cumpliendo una condición para que el bucle se siga ejecutando n veces
# Vamos a calcular la raíz cuadrada de un número, este debe de ser estrictamente positivo
import math
numero = int(input('Digite un número: '))
while numero < 0: # Mientras el usuario siga ingresando un número menor a 0 el bucle while se sigue ejecutando 
    print('Error -> Debe digitar un número positivo estrictamente')
    numero = int(input('Digite un número: '))

# Si el usuario ingresa un número mayor o igual que 0 el bucle While NO se ejecuta
print(f'\nLa raíz cuadrada de {numero} es {(math.sqrt(numero)):.2f}\n') 

# Podemos ocuparlo también para un número detemrinado de iteraciones con una variable iteradora
# Vamos a imprimir un 'Hola Mundo' 10 veces
i = 1
while i <= 10:
    print(f'Hola Mundo {i}')
    i += 1 # Incrementamos el contador en una unidad
```

```python
'''
38. programación en Python | Bucle For
'''
# Es un bucle con un número determinado de iteraciones
# Es decir, a priori, nosotros conocemos el número de veces que se va a repetir
# El bucle for en Python se puede utilizar con colecciones
# Vamos a tener un iterador que va a recorrer los elementos de una colección (tuplas, arreglos, listas, etc.)
for i in [1,2,3]:
    print('Hola Mundo') # Imprime 'Hola Mundo' 3 veces debido a que recorre la colección elemento por elemento
for i in [1,2,3,'Eduardo',10]: # NO importa el orden o tipo de elemento, sino el número de elementos en la colección
    print('Hola Mundo') # Imprime 'Hola Mundo' 5 veces debido a que recorre la colección elemento por elemento
for i in [1,2,3,'Eduardo',10]:
    print(f'Elemento: {i}') # Imprime cada elemento a evaluar por la iteración correspondiente
# Vamos a ver cómo trabajar el bucle for con diccionarios {clave:valor} 
coleccion = {'Eduardo':29,'Kenya':30,'David':18,'Laura':24}
for i in coleccion:
    print(f'Elemento {i}') # Esto nos va a mostrar únicamente las claves del diccionario
for i in coleccion:
    print(coleccion[i]) # Esto nos va a mostrar únicamente los valores del diccionario
for i in coleccion:
    print(f'{i} -> {coleccion[i]}') # Esto nos va a mostrar tanto clave como valor
for clave,valor in coleccion.items(): # utilizamos el método .items()
    print(f'{clave} -> {valor}') # Este método es un poco más profesional que el anterior
cadena = 'Eduardo'
for i in cadena: # Va a recorrer carácter por carácter de la cadena
    print(f'{i}')
for i in cadena: # Va a recorrer carácter por carácter de la cadena
    print(f'{i}', end='') # Con el parámetro end='' hace que se imprima en la misma línea todo el recorrido de la cadena
```

```python
'''
39. Programación en Python | Bucle For tipo range
'''
# Podemos utiliza este tipo range para que el bucle for se repita un número de veces de manera sencilla
# Hay que recordar que el contador de Python siempre comienza en 0 si no se especifica lo contrario
for i in range(10): # Esto crea una colección 'ficticia' con un número de elementos 'i'
    print(f'Vamos en el elemento: {i}')
for i in range(15,20): # Esto crea una colección 'ficticia' dentro del rango 5 a 10 (siempre llega a n-1)
    print(f'Vamos en el elemento: {i}')
for i in range(20,30,2): # Esto crea una colección 'ficticia' dentro del rango 20 a 30 pero con saltos de 2 en 2 
    print(f'Vamos en el elemento: {i}')
for i in range(70,60,-2): # Esto crea una colección 'ficticia' dentro del rango 70 a 60 pero con saltos de 2 en 2
    print(f'Vamos en el elemento: {i}')
```

```python
'''
40. Programación en Python | Instrucción continue y break
'''
# Estas instrucciones pueden ocuparse en cualquiera de los dos bucles (for & while)
# Vamos a explicarlo en el bucle for tipo range
for i in range(10):
    if i == 5 or i == 6:
        continue # esta instrucción obvia el elemento y se lo salta, continuando con el siguiente
    print(i)
for i in range(10):
    if i == 5: # Si la iteración es igual al elemento 5 se entra en este if
        break # esta instrucción rompe el bucle for sin importar si faltan más iteraciones
    print(i)
print('Programa finalizado')
```

```python
'''
41. Programación en Python | Ejercicio 1
'''
# Llenar una lista con los números del 1 al 50, luego mostrar la lista con un bucle for
# Los elementos deben mostrarse de la siguiente forma: 1-2-3-4-5-...-49-50
lista = []
for i in range(51):
    lista.append(i)
for j in lista:
    print(j,end='-')
# Otra manera de realizarlo es declarar desde un principio la lista en vez de bucle
print('\n')
lista = list(range(0,51))
for k in lista:
    print(k,end='-')
```

```python
'''
42. Programación en Python | Ejercicio 2
'''
# Llenar una lista con los números del 1 al 10, luego modificar los elementos de la lista
# La modificación será multiplicarlos por un valor que el usuario digite
lista = list(range(1,11))
for i in lista:
    print(i, end=',')
while True:
    try:
        valor = int(input('\n¿Por cuál número desea multiplicar los elementos?: '))
    except ValueError:
        print('Ingresa un número válido, debe ser un número entero')
        continue
    else:
        break
indice = 0
for i in lista:
    lista[indice] *= valor
    indice += 1
print(f'\nLista final con los elementos multplicados por {valor}:')
for k in lista:
    print(k, end=',')
```

```python
'''
43. Programación en Python | Ejercicio 3
'''
# Pide número e ingresalos en una lista, cuando el usuario ingrese un 0 dejamos de agregar nuevos elementos
# Por último muestra, los números de menor a mayor
lista = []
salida = False
while not salida:
    while True:
        try: # Intentamos el casting de la cadena ingresada
            entrada = int(input('¿Qué número desea agregar a la lista?: '))
        except ValueError: # Si el casting no es posible iteramos de nuevo
            print('Ingrese n número válido, debe ser un número entero')
            continue
        else: # Si el casting fue posible imprimimos y salimos del ciclo
            break
    if entrada == 0:
        salida = True
    else:
        lista.append(entrada)
lista.sort()
print(f'La lista ordenada sería: {lista}')
```

```python
'''
44. Programación en Python | Ejercicio 4
'''
# Hacer un programa para sumar números pares dentro de un rango
a = int(input('¿De dónde va a comenzar a sumar?: '))
b = int(input('¿Hasta dónde quiere comenzar a sumar?: '))
suma = 0
for i in range(a,b+1):
    if i%2 == 0:
        suma += i
print(f'La suma es {suma}')
```

```python
'''
45. Programación en Python | Ejercicio 5
'''
# Hacer un programa para calcular el factorial de un número positivo
while True:
    try:
        numero = int(input('Digite un número: '))
        while numero < 0:
            print('Error -> El número debe ser mayor que 0')
            numero = int(input('Digite un número: '))
    except ValueError:
        print('Ingrese un número válido, debe ser un número entero')
        continue
    else:
        break
factorial = 1
for i in range(1,numero+1): # Recordemos que el rango se evalua siempre hasta n-1 
    factorial *= i # Usamos la multiplicación en asignación para calcular el factorial
print(f'El factorial del {numero} es igual a {factorial}')
```

```python
'''
47. Programación en Python | Ejercicio 7
'''
# Realiza un juego para adivinar un número, para ello generar un número aleatorio N entre 0 y 100
# Luego ir solicitando númerps indicando 'es mayor' o 'es menor' según sea el caso con respecto al número N
# El juego termina cuando el usuario acierta y muestra el número de intentos
import random
print('\t.:Juego -Adivina el número:.')
aleatorio = random.randint(0,100) # Generamos un número pseudo aleatorio
contador = 0
while True:
    numero = int(input('Digite un número: '))
    contador += 1
    if numero > aleatorio:
        print('\tNo es el número, digite un número menor')
    elif numero < aleatorio:
        print('\tNo es el número, digite un número mayor')
    else:
        print(f'\t¡Felicidades! Acabas de adivinar el número aleatorio -> {aleatorio}')
        break
print(f'\nNúmero de intentos: {contador}')
```

```python
'''
48. Programación en Python | Ejercicio 8
'''
# Un programa que simule un cajero automático con un saldo incial de $1000 y con el siguiente menú de opciones:
# 1. Ingresar dinero en la cuenta
# 2. Retirar dinero de la cuenta
# 3. Mostrar dinero disponible
# 4. Salir
saldo = 1000
while True:
    print('\t .:Menú:.')
    print('1. Ingresar dinero en la cuenta')
    print('2. Retirar dinero de la cuenta')
    print('3. Mostrar dinero disponible')
    print('4. Salir')

    opcion = int(input('Digite una opción: '))
    if opcion == 1:
        extra = float(input('¿Cuánto dinero desea ingresar a la cuenta?'))
        saldo += extra
        print(f'El aldo disponible en la cuenta es de ${saldo}')
    elif opcion == 2:
        retiro = float(input('¿Cuánto dinero desea retirar de la cuenta?'))
        if retiro > saldo:
            print('No tiene ese saldo disponible')
        else:
            saldo -= retiro
            print(f'El saldo disponible en la cuenta es de {saldo}')
    elif opcion == 3:
        print(f'El saldo disponible en la cuenta es de {saldo}')
    elif opcion == 4:
        print(f'Gracias por ocupar este cajero automático')
        break # La opción 4 de salida ejecuta un break para finalizar el programa
    else: # Mientras la opción sea inválida o distinta de 4, no se ejecuta ningún 'break' por lo tanto el bucle while se sigue ejecutando
        print(f'Opción inválida')
```

```python
'''
49. Programación en Python | Ejercicio 9
'''
# Hacer un programa donde el usuario ingrese una frase cualquiera
# Dicha frase le será devuelta sin espacios y contando el número de caracteres (sin contar espacios en blanco)
frase = input('Escriba una frase: ')
frase2 = ''
for i in frase: # Iteramos sobre cada elemento de la frase almacenada
    if i != ' ':
        frase2 += i # Agregamos cada elemento iterado siempre que no sea un espacio en blanco
frase = frase2
print(f'Frase final es: {frase}')
print(f'Número de caracteres: {len(frase)}')
```

```python
'''
50. Programación en Python | Ejercicio 10
'''
# Hacer un programa que pida una cadena por teclado
# Luego meta los caracteres en una lista sin repetir caracteres
cadena = input('Digite una cadena: ')
lista = []
for i in cadena: # Iteramos sobre cada elemento de la cadena almacenada
    if i not in lista:
        lista.append(i) # Agregamos cada elemento iterado siempre que no esté previamente agregado (repetido)
print(f'\nLista de caracteres sin repetir: \n{lista}')
```

```python
'''
51. Programación en Python | Ejercicio 11
'''
# Hacer un prorama que simule una agenda de contactos mediante el uso de diccionarios
# Que la clave sea el nombre del usuario y el valor sea su número telefónico
#  También que el programa tenga el siguiente menú de opciones:
#  1. Nuevo contacto
#  2. Borrar contacto
#  3. Ver contactos existentes
#  4. Salir
agenda = {}
while True:
    print('\t.:MENU:.')
    print('1. Nuevo contacto')
    print('2. Borrar contacto')
    print('3. Ver contactos existentes')
    print('4. Salir')
    opcion = int(input('Digite una opción del menú: '))

    print()

    if opcion == 1:
        nombre = input('Nombre del contacto: ')
        telefono = input('Número de teléfono: ')

        if nombre not in agenda:
            agenda[nombre] = telefono
            print('¡Contacto agregado!')
        else:
            print('Ese nombre de contacto ya existe')
    elif opcion == 2:
        nombre = input('Nombre del contacto: ')

        if nombre in agenda:
            del(agenda[nombre])
            print('¡Contacto eliminado')
        else:
            print('Ese contacto no existe')

    elif opcion == 3:
        print('Agenda de contactos: ')
        for clave,valor in agenda.items():
            print(f'Nombre: {clave}, Teléfono: {valor}')
    
    elif opcion == 4:
        print(f'Gracias por utilizar su Agenda de Contactos')
        break # La opción 4 de salida ejecuta un break para finalizar el programa

    else: # Mientras la opción sea inválida o distinta de 4, no se ejecuta ningún 'break' por lo tanto el bucle while se sigue ejecutando
        print('Se equivocó de opción de menú')

    print()
```
