```python
'''
52. Programación en Python | Cadena de caracteres
'''
# Podemos crear una cadena simplemente poniéndolo entre comillas simples o dobles
cadena = 'Hola Mundo'
print(cadena)
# Podemos incluso una combinar las comillas simples con las dobles
cadena = 'Estoy "estudiando"'
print(cadena)
# Para que podamos ocupar dos veces las mismas comillas dentro de una cadena ocupamos \ para diferenciar
cadena = "Estoy \"estudiando\""
print(cadena)
# Para escribir una tabulación ocuparemos \t, el cuál nos dará 4 espacios
cadena = 'Estoy \testudiando'
print(cadena)
# Para escribir un salto de línea ocuparemos \n, el cuál nos brinca exactamente a la siguiente línea
cadena = 'Estoy \nestudiando'
print(cadena)
# Para guardar una cadena 'en crudo', es decir, sin saltos de línea ni tabulaciones ocuparemos 'r' antes de la cadena
cadena = (r"D:\nombre\trabajos")
print(cadena)
# Para guardar cadenas multílinea ocuparemos 3 comillas de cada lado
cadena = '''
Mi nombre es Eduardo
Tengo 29 años
¿Cuál es tu nombre y edad?
'''
print(cadena)

cadena1 = 'Hola'
cadena2 = ' ¿Cómo estás?'
print(cadena1 + cadena2) # Podemos sumar ambas cadenas para unirlas
```

```python
'''
53. Programación en Python | Índices y Slicing
'''
# Si queremos mostrar sólo un elemento específico de la cadena, lo haríamos igual que con otras colecciones
# Es decir, ocupando los índices dependiendo del elemento que mostraremos
cadena = 'Eduardo'
print(cadena)
print(cadena[0])
# En las cadenas también se cumplen los índices negativos como en las demás colecciones
print(cadena[-3])
# También podemos hacer slicing en las cadenas de caracteres
print(cadena[1:4]) # Nos muestra desde el índice 1 hasta el elemento n-1, en este caso, el elemento 3
# Las cadenas de caracteres son inmutables, es decir, no hay métodos para modificarlas
# Podríamos modificarlas sobreescribiendo la cadena o reemplazando parte de la cadena ya existente
cadena = 'e' + cadena[1:]
print(cadena)
# También podemos ocupar el método len() en las cadenas de caracteres
print(len(cadena)) # Nos muestra la longitud de la cadena, en este caso, son 7 elementos
```

```python
'''
54. Programación en Python | Métodos para cadenas (Parte 1)
'''
cadena = 'hola mundo'.upper() # El método .upper() transforma toda la cadena en mayúsculas
print(cadena)
cadena = 'HOLA mundo'.lower() # El método .lower() transforma toda la cadena en minúsculas
print(cadena)
cadena = 'hola mundo'.capitalize() # El método .capitalize() transforma la primera letra en mayúscula
print(cadena)
cadena = 'hola mundo'.title() # El método .title() transforma las primeras letras en mayúsculas
print(cadena)
cadena = 'hola mundo'.count('o') # El método .count('x') cuenta el número de elementos 'x' en la cadena
print(cadena)
cadena = 'hola mundo'.find('a') # El método .find('x') muestra el primer índice donde encuentra el elemento 'x'
print(cadena)
cadena = 'hola mundo'.rfind('o') # El método .rfind('x') muestra el último índice donde encuentra el elemento 'x'
print(cadena)
# Podemos averiguar si la cadena está compuesta únicamente por valores numéricos
cadena = '1000'.isdigit() # El método .isdigit() nos dice su la cadena está o no compuesta sólo por dígitos
print(cadena)
# Podemos averiguar si la cadena está compuesta únicamente por valores alfabéticos
cadena = 'abcd'.isalpha() # El método .isalpha() nos dice si la cadena está o no compuesta sólo por letras
print(cadena)
# Podemos averiguar si la cadena está compuesta únicamente por valores alfanuméricos (tanto letras como números)
cadena = 'ab12'.isalnum()
print(cadena) # El método .isalnum() nos dice si está o no compuesta sólo por caracteres alfanuméricos
# Podemos averiguar si la cadena está compuesta únicamente por letras mayúsculas
cadena = 'HOLA MUNDO'.isupper() # El método .isupper() nos dice si la cadena está o no compuesta sólo por letras mayúsculas
print(cadena)
# Podemos averiguar si la cadena está compuesta únicamente por letras mayúsculas
cadena = 'hola mundo'.islower() # El método .islower() nos dice si la cadena está o no compuesta sólo por letras minúsculas
print(cadena)
# Podemos averiguar si la cadena está compuesta por un título, es decir, las primeras letras en mayúscula
cadena = 'Hola Mundo'.istitle()  # El método .istitle() nos dice si la cadena está o no compuesta por un título
print(cadena)
# Podemos averiguar si toda la cadena está compuesta por espacios en blanco
cadena = '      '.isspace() # El método .islower() nos dice si la cadena está o no compuesta sólo por espacios en blanco
print(cadena)
```

```python
'''
55. programación en Python | Métodos para cadenas (Parte 2)
'''
# Podemos averiguar si la cadena comienza con cierto carácter o caracteres específicos usando .startswith()
cadena = 'hola mundo'.startswith('hola')
print(cadena)
# Podemos averiguar si la cadena finaliza con cierto carácter o caracteres específicos usando .endswith()
cadena = 'hola mundo'.endswith('mundo')
print(cadena)
# Podemos separar los elementos de una cadena para que se conviertan en una lista
cadena = 'hola mundo'.split() # El método .split() separa los elementos cada que encuentra un espacio en blanco
print(cadena)
# Podemos especificar al método .split('x') para separar la cadena cada que encuentra el caracter 'x'
cadena = 'hola-mundo-bonito'.split('-')
print(cadena)
# Podemos utilizar el método .join(x) para separar los elementos de una cadena 'x' mediante un carácter específico
cadena = '|'.join('eduardo')
print(cadena)
# Podemos eliminar los espacios en blanco de una cadena usando el método .strip()
cadena = '        eduardo        '.strip()
print(cadena)
# podemos eliminar ciertos caracteres 'x' innecesarios en las cadenas, con el método .strip(x)
cadena = '..........eduardo..........'.strip('.')
print(cadena)
# Podemos reemplazar ciertos elementos de la cadena por otro elemento específico usando el método replace(x,y)
cadena = 'hola MUNDO'.replace('hola','HOLA')
print(cadena)
```

```python
'''
56. Programación en Python | Ejercicio 1
'''
# Hacer un programa donde se va a imprimir la cadena con más caracteres de dos palabras dadas
# En caso de tener el mismo número de caracteres imprimir "Tienen los mismo caracteres"
cadena1 = input('Digite una cadena: ')
cadena2 = input('Digite otra cadena: ')
if len(cadena1) > len(cadena2): # Comparamos las longitudes de las cadenas
    print(f'La cadena más larga es: {cadena1}')
elif len(cadena2) > len(cadena1): # Comparamos las longitudes de las cadenas
    print(f'La cadena más larga es: {cadena2}')
else:
    print(f'Las cadenas {cadena1} & {cadena2} tienen los mismos caracteres')
```

```python
'''
57. Programación en Python | Ejercicio 2
'''
# Programa para detectar si una frase introducida por el usuario finaliza o no con un punto '.'
# Imprimir 'Termina con un punto' o 'No termina con un punto' dependiendo del caso
cadena = input('Digite una cadena: ')
if cadena.endswith('.') == True: # Evaluamos si termina o no en un punto (.)
    print('La cadena finaliza con un punto')
else:
    print('La cadena no finaliza con un punto')
```

```python
'''
58. Programación en Python | Ejercicio 3
'''
# Hacer un programa que pida una cadena al usuario y detecte si la frase o palabra es un palíndromo
cadena = input('Digite una cadena: ')
cadena_aux = cadena.replace(' ','')
cadena_aux = cadena_aux.lower()
cadena2 = cadena_aux[::-1] # Invertimos el orden de la cadena
print(f'La cadena invertida es {cadena2}')
if cadena_aux == cadena2: # Comparamos si la cadena invertida es igual a la cadena original sin espacios
    print(f'La cadena "{cadena}" es un palíndromo')
else:
    print(f'La cadena "{cadena}" no es un palíndromo')
```

```python
'''
59. Programación en Python | Ejercicio 4
'''
# Hacer un programa donde se reemplacen todos los espacios en blanco por asteriscos '*'
# Y además que cada palabra comience por mayúsculas y las demás letras sean minúsculas
cadena = input('Digite una cadena: ')
cadena = cadena.replace(' ', '*') # Reemplazamos los espacios por *
cadena = cadena.title() # Reemplazamos las letras iniciales por mayúsculas
print(cadena)
```
