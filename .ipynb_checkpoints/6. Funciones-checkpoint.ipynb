```python
'''
61. Programación en Python | Funciones sin retorno de valor
'''
# Comenzamos las funciones con la palabra reservada 'def'
def saludar(nombre):
    print(f'Hola {nombre}')

saludar('Eduardo') # Llamamos a la función que hemos creado anteriormente con 1 solo parámetro

def tabla_multiplicar(num):
    for i in range(1,11):
        print(f'{num}*{i} = {num*i}')

tabla_multiplicar(5) # Llamamos a la función con el parámetro 5
# Estas funciones que hemos presentado son sin retorno de valor, sólo devuelven mensajes
```

```python
'''
62. Programación en Python | Funciones con retorno de valor
'''
# Las funciones con retorno de valor se comienza con el comando 'def'
def multiplicar(num1,num2):
    mult = num1*num2
    return mult # Aquí nos regresa el valor de la multiplicación de num1*num2
resultado = multiplicar(105,378) # Llamamos a la función declarando los dos parámetros que elegimos
print(resultado)

def prueba(): # Esta función únicamente nos regresa 3 elementos sin procesar nada más
    return 'Hola', 45, [12,13,17]
c,n,l = prueba()
print(c)
print(n)
print(l)
```

```python
'''
63. Programación en Python | Argumentos y parámetros
'''
# Hay que aprender a distinguir entre argumentos y parámetros porque son conceptos distintos
def resta(num1,num2): # Al crear la función, los valores entre paréntesis son los parámetros
    return num1 - num2
print(resta(7,3)) # Al llamar la función, los valores que le damos son argumentos
# Si no queremos mandar argumentos por posición sino por nombre debemos especificarlo
print(resta(num2=7,num1=3)) # Estos argumentos fueron mandados por nombre y NO por posición 
```

```python
'''
64. Programación en Python | Argumentos por valor o por referencia
'''
def doblar_valor(numero):
    numero *= 2
    return numero
n = 5
doblar_valor(5) # Le estamos pasando sólo una copia, el valor original de n NO se ha modificado
print(f'{n}, {doblar_valor(5)}') # Esto es pasar el argumento por valor y NO por referencia

# Todas las colecciones en Python se pasan con argumentos por referencia
def doblar_valores(numeros):
    for i,n in enumerate(numeros):
        numeros[i] *= 2
n = [5,10,15,20]
doblar_valores(n) # Cualquier modificación que se haga dentro de la función afecta a la variable externa
print(n) # Las colecciones al mandar argumentos por referencia sí son modificadas

# Si queremos mandar los argumentos por valor y no por referencia, para que la colección no se modifique
# Vamos a mandar el valor en vez de la colección completa
doblar_valores(n[:])
print(n) # En este caso ya no se afectó la lista
```

```python
'''
65. Programación en Python | Ejercicio 1
'''
# Desarrollar un programa que pueda calcular el valor del tipo de cambio de una moneda contra el dólar
# Es decir, el tipo de cambio de tu moneda hacía el dólar y viceversa
def cambio_pesos_a_dolares(pesos):
    return pesos/18.8

def cambio_dolares_a_pesos(dolares):
    return dolares*18.8

while True:
    print('''\t.:Menú:.
          1. Convertir pesos a dólares
          2. Convertir dólares a pesos
          3. Salir''')

    opcion = int(input('Digite una opción del menú: '))
    print()
    if opcion == 1:
        pesos = float(input('Digite la cantidad de pesos: '))
        print(f'Dólares -> ${cambio_pesos_a_dolares(pesos):.2f}')
        print()
    elif opcion == 2:
        dolares = float(input('Digite la cantidad de dólares: '))
        print(f'Pesos -> ${cambio_dolares_a_pesos(dolares):.2f}')
        print()
    elif opcion == 3:
        break
    else:
        print(f'Se equivocó de opción de menú')
        print()
```

```python
'''
66. Programación en Python | Ejercicio 2
'''
# Hacer un programa que pida la anchura y altura de un rectángulo
# Y que con ayuda de una función dibuje el rectángulo con '*'
ancho = int(input('Digite la anchura: '))
alto = int(input('Digite la altura: '))
def dibujar(ancho,alto):
    for i in range(alto):
        for j in range(ancho):
            print('*',end='')
        print()
dibujar(ancho,alto)
```

```python
'''
67. Programación en Python | Ejercicio 3
'''
# Crear un programa que tenga una lista de clientes, cada cliente debe tener Nombre, Apellido y DNI
# El programa tendrá la siguiente lista de opciones:
# 1.- Agregar nuevo cliente
# 2.- Mostrar todos los clientes
# 3.- Mostrar clientes por DNI
# 4.- Eliminar cliente
# 5.- Salir
def agregar_cliente(clientes,nombre,apellido,dni):
    cliente = {}
    cliente['nombre'] = nombre
    cliente['apellido'] = apellido
    cliente['dni'] = dni
    clientes.append(cliente)

def mostrar_clientes(clientes):
    for i in clientes:
        print(f'Nombre: {i['nombre']}, Apellido: {i['apellido']}, DNI: {i['dni']}')

def mostrar_cliente(clientes,dni):
    for i in clientes:
        if i['dni'] == dni:
            print(f'Nombre: {i['nombre']}, Apellido: {i['apellido']}, DNI: {i['dni']}')
            return True
    return False
    
def eliminar_cliente(clientes, dni):
    for i in clientes:
        if i['dni'] == dni:
            clientes.remove(i)
            return True
    return False
    
clientes = [] # Creamos una lista

while True:
    print('''\t.:MENU:.
          1. Agregar nuevo cliente
          2. Mostrar todos los clientes
          3. Mostrar clientes por DNI
          4. Eliminar cliente
          5. Salir''')
    opcion = int(input('Digite una opción: '))
    print()
    if opcion == 1:
        nombre = input('Nombre del cliente -> ')
        apellido = input('Apellido del cliente -> ')
        dni = input('DNI del cliente -> ')
        agregar_cliente(clientes,nombre,apellido,dni)
    elif opcion == 2:
        mostrar_clientes(clientes)
    elif opcion == 3:
        dni = input('DNI del cliente -> ')
        if mostrar_cliente(clientes,dni):
            print('Cliente encontrado')
        else:
            print('Cliente no encontrado')
    elif opcion == 4:
        dni = input('DNI del cliente -> ')
        if eliminar_cliente(clientes,dni):
            print('Cliente eliminado')
        else:
            print('Cliente no encontrado')
    elif opcion == 5:
        break
    else:
        print('Error, se equivocó de opción de menú')
    print()
```

```python
'''
68. Programación en Python | Funciones Recursivas
'''
# Las funciones recursivas son un caso similar a los ciclos, es decir, eventualmente tienen un final
# Son iterativas, deben de tener un caso base
def cuenta_regresiva(num):
    if num > 0:
        print(num)
        cuenta_regresiva(num-1) # Ocupamos la misma función iterando y reduciendo el contador (num)
    else:
        print('¡BOOM! La bomba explotó')
cuenta_regresiva(10) # Inicializamos la función con un contador = 10
```

```python
'''
69. Programación en Python | Ejercicio 4
'''
# Desarrollar un programa para calcular el factorial de un número con ayuda de una función recursiva
def factorial(num):
    if num > 0:
        resultado = num * factorial(num-1)
    else:
        resultado = 1
    return resultado
valor = factorial(6)
print(valor)
```

```python
'''
70. Programación en Python | Ejercicio 5
'''
# Desarrollar un programa que permita sumar los dígitos de un número con ayuda de una función recursiva
def SumarDigitos(num):
    if num == 0:
        resultado = 0
    else:
        resultado = SumarDigitos(int(num/10)) + (num%10)
    return resultado
valor = SumarDigitos(1694714543541542)
print(valor)
```
