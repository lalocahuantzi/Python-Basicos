```python
'''
25. Programación en Python | Listas (Parte 1)
'''
# Son algo parecido a lo que se conoce en otros lenguajes como arreglos o vectores
# Podemos almacenar valores booleanos, alfanuméricos, cadenas e incluso otras listas
# Las listas se denotan por [] y se separan por ','

lista = ['Lunes', 'Martes', 'Miércoles', 'Jueves', 'Viernes']
# Mediante print podemos imprimir todos los elemtnso de la lista
print(lista)
# Podemos indexar el elemento que queremos mostrar
# Los índices en Python siempre van desde el índice 0 hasta el índice n-1
print(lista[0])
# Los índices también pueden ser accedidos desde atrás hacía delante
# El índice -1 siempre nos muestra el último elemento que tiene la lista
print(lista[-1])
# Podemos imprimir únicamente una parte de la lista limitando los índices
# Siempre llega hasta una posición antes del límite superior
print(lista[0:3])
# Los ':' significan desde el principio o hasta el final de la lista, dependiendo de dónde son colocados
print(lista[:2])
print(lista[2:])
```

```python
'''
26. Programación en Python | Listas (Parte 2)
'''
# El método 'len(lista)' nos devuelve el número de elementos que componen nuestra lista
lista = ['Lunes','Martes','Miercoles','Jueves','Viernes', 40,5.67,[1,2,3],True]
print(len(lista))
# Vamos a crear una nueva lista para aprender a agregar o eliminar elementos de ella
lista = [1,2,3,4,5]
lista.append(6)
print(lista)
# Vamos a crear una nueva lista para aprender a agregar o eliminar elementos de la lista
lista = [1,2,4,5] # A esta nueva lista le falta el elemento '3'
lista.insert(2,3) # Con el método 'lista.insert(i, x)' agregas el elemento x en la posición i de la lista
print(lista)
# Si quieres agregar más de un elemento a la lista ocupamos el método 'lista.extend(x,y,z)'
lista.extend([6,7,8]) # La nueva lista será agregada al final de la lista original
print(lista)
# Vamos crear 2 listas nuevas para demostrar la concatenación al sumar 2 listas diferentes
lista1 = [1,2,3,4,5]
lista2 = [6,7,8]
lista3 = lista1 + lista2
print(lista3)
# Creamos una nueva lista para mostrar la búsqueda  de elementos dentro de las listas
lista = [1,2,3,4,5,'Eduardo']
print('Eduardo' in lista) # Con el operador 'in' se valida si un elemento pertenece a la lista, nos devuelve True/False
print(lista.index(5)) # El método 'lista.index(x)' nos devuelve el índice en el que se encuentra el elemento x
# Creamos una nueva lista para contar el número de elementos x en una lista
lista = [1,2,3,4,5,6,1,2,7,4,5,6,2,3,4]
print(lista.count(5)) # El método lista.count(x) nos devuelve el número de elementos x en la lista
# Creamos una nueva lista para aprender a eliminar elementos de una lista
lista = [1,2,3,4,5,'Eduardo']
lista.pop() # El método lista.pop() sin argumento dentro del parentsis elimina el último elemento de la lista
print(lista)
lista = [1,2,3,4,5]
lista.pop(3) # El método lista.pop(i) elimina el elemento con índice i de la lista
print(lista)
lista = [1,2,3,4,5,'Eduardo']
lista.remove('Eduardo') # Podemos eliminar el elemento con valor x con el método lista.remove(x)
print(lista)
lista.clear() # El métodon lista.clear() nos elimina todos los elementos de la lista
print(lista)
# Vamos a crear una nueva lista para mostrar distintas funciones
lista = [1,2,3,4,5,'Eduardo']
lista.reverse() # El método lista.reverse() invierte todos los elementos de la lista
print(lista)
lista = [1,2,3,4,5,'Eduardo']*2 # Al multiplicar x2, duplicamos todos los elementos de la lista
print(lista)
lista = [5,-7,8,3,-5,10,17]
lista.sort() # El método lista.sort() nos ordena los elementos de menor a mayor valor
print(lista)
lista = [5,-7,8,3,-5,10,17]
lista.sort(reverse=True) # Con el argumento 'reverse=True' se ordenan los elementos de mayor a menor valor
print(lista)
```

```python
'''
27. Programación en Python | Tuplas
'''
# Son otro tipo de colecciones de características similares a las listas
# La gran diferencia es que las tuplas son inmutables y sirven mucho para hacer búsquedas
# Son mucho más rápidas al momento de ejecución y consumen mucho menos memoria que las listas
tupla = (4,7.2,'Hola',9.1,True,7.2,False,[1,2,3],7.2)
print(tupla) # Podemos mostrar los elementos de la tupla exactamente igual que las listas
print(tupla[-1]) # Podemos mostrar el elemento con índice i exactamente igual que en las listas
print(tupla[2:]) # Podemos hacer el 'slicing' exactamente igual que en las listas
print(tupla.index('Hola')) # Podemos localizar el índice del elemento x con tupla.index(x)
print(tupla.count(7.2)) # Podemos contar el número de elementos x con tupla.count(x)
# No podemos ocupar funciones como .append() o .remove() entre otras modificaciones que las listas sí permiten
# Podemos convertir las tuplas en listas con la función list(tupla)
lista = list(tupla)
print(lista)
# Podemos convertir las listas en tuplas con la función tuple(lista)
lista = [1,2,3,4,5,'Eduardo']
tupla = tuple(lista)
print(tupla)
```

```python
'''
28. Programación en Python | Conjuntos (Parte 1)
'''
# Son un tipo de colección donde los elementos se agregan de forma desordenada
# Su principal característica es que no puede haber elementos duplicados
# El problema es que los conjuntos y los diccionarios comparte el símbolo {}
conjunto = set() # Con la función set() indicamos que es un conjunto y NO un diccionario
conjunto = {} # Conjunto vacío
print(conjunto)
conjunto = {1,2,3,'Hola'} # No puedes agregar otro tipo de colección dentro (listas, tuplas, arreglos, etc.)
print(conjunto)
conjunto = {1,2,3,'Hola',1,2,3} # En el conjunto NO puede haber elementos duplicados
print(conjunto)
conjunto.add(4) # Con el método conjunto.add(x) podemos agregar un elemento x al conjunto
print(conjunto) # El elemento NO se agrega al final, el conjunto lo agregará donde él decida colocarlo
conjunto.discard('Hola') # Con el método conjunto.discard(x) podemos eliminar el elemento x del conjunto
print(conjunto)
conjunto.clear() # Con el método conjunto.clear() podemos eliminar todos los elementos del conjunto
print(conjunto) # Únicamente nos quedaría set()
# Vamos a crear un nuevo conjunto para seguir mostrando funciones
conjunto = {1,2,3,'Hola',4.56}
print(3 in conjunto) # Podemos corroborar si el elemento 3 pertenece  al conjunto, devuelve True/False
print(2 not in conjunto) # Podemos corroborar si el elemento 2 NO pertenece al conjunto, devuelve True/False
```

```python
'''
29. Programación en Python | Conjuntos (Parte 2)
'''
# Vamos a realizar las operaciones entre conjuntos que se conocen en Matemáticas
conjunto_a = {1,2,3}
conjunto_b = {3,4,5}
# Podemos comprobar si el conjunto a es igual al conjunto b, es decir, tienen los mismos elementos
print(conjunto_a == conjunto_b)
# Vamos a realizar la unión de ambos conjuntos
conjunto_c = conjunto_a | conjunto_b # La unión se realiza con el símbolo '|'
print(conjunto_c)
conjunto_c = conjunto_a & conjunto_b # La intersección se realizar con el símbolo '&'
print(conjunto_c)
conjunto_c = conjunto_a - conjunto_b # La diferencia se realiza '-'
print(conjunto_c) # Nos muestra elementos que están en el conjunto_a pero no están en el conjunto_b
conjunto_c = conjunto_b - conjunto_a # La diferencia se realiza con '-'
print(conjunto_c) # Nos muestra elementos que están en el conjunto_b pero no están en el conjunto_a
conjunto_c = conjunto_a ^ conjunto_b # La diferencia simétrica se realiza con '^'
print(conjunto_c) # Nos muestra elementos que están en el conjunto_a y en el conjunto_b pero NO en ambos

conjunto_c = {1,2,3,4,5}
# Verificamos si conjunto_a es subconjunto del conjunto_c
print(conjunto_a.issubset(conjunto_c))
# Verificamos si conjunto_c es subconjunto del conjunto_a
print(conjunto_c.issubset(conjunto_a))
# Verificamos si conjunto_a es disconjunto del conjunto_b, el resultado es FALSE debido a que comparten el '3'
print(conjunto_a.isdisjoint(conjunto_b))
# Para obtener conjuntos inmutables ocupamos la función frozenset()
conjunto_d = frozenset({1,2,3}) # No se permite ni añadir ni eliminar elementos de este conjunto
```

```python
'''
30. Programación en Python | Diccionarios (Parte 1)
'''
# Son otro tipo de colecciones donde los elementos también están desordenados
# La gran diferencia es que poseen 2 elementos por posición, la clave y el valor
# Las claves NO pueden estar duplicadas, los valores sí lo pueden estar
# La forma estándar del diccionario es {clave:valor}
# Haremos un diccionario para saber cómo decir los colores en inglés
diccionario = {'azul':'blue','rojo':'red','verde':'green'}
print(diccionario) # Aquí nos muestra todos los elementos del diccionario
print(diccionario['azul']) # Podemos buscar elementos por su clave, en este caso la clave es 'azul'
# Para agregar nuevos elementos será por su clave y lo igualamos a su valor
diccionario['amarillo'] = 'yellow' # En este caso la clave es 'amarillo' y el valor es 'yellow'
print(diccionario)
# Podemos modificar el diccionario especificando la clave y su nuevo valor
diccionario['azul'] = 'BLUE'
print(diccionario)
# Para eliminar elementos del diccionario ocupamos la palabra reservada 'del(clave)'
del(diccionario['azul']) # Al eliminar una clave se eliminan por completo sus valores
print(diccionario)
# Dentro de los diccionarios podemos ingresar números enteros, flotantes, cadenas, booleanos
# Podemos incluso agregar otras colecciones como listas, tuplas, conjuntos e incluso otros diccionarios
diccionario = {'Eduardo':[29,165],'Kenya':[30,150]}
print(diccionario['Eduardo'])
```

```python
'''
31. Programación en Python | Diccionarios (Parte 2)
'''
equipo = {10:'Lionel Messi',7:'Cristiano Ronaldo',14:'Thierry Henry',9:'Ronaldo Nazario'}
print(equipo[10]) # Buscamos el valor con clave '10'
# En caso de no saber si existe una clave podemos ocupar el método .get(clave, en caso de no encontrarlo)
print(equipo.get(6,'No hay un jugador registrado con ese dorsal')) 
# Podemos corroborar dentro de un diccionario exactamente igual que en las demás colecciones
print(10 in equipo) # Nos regresa un valor booleano dependiendo de si existe la clave o no en el diccionario
# Si queremos que nos muestre todas las claves del diccionario utilizamos el método diccionario.keys()
print(equipo.keys()) # En este caso, nos mostrará todos los dorsales que existen en este equipo
# Si queremos que nos muestre todos los valores del diccionario utilizamos el método diccionario.values()
print(equipo.values()) # En este caso, nos mostrará todos los jugadores que existen en este equipo
# El método para recorrer diccionarios con el bucle 'for' ocupamos el método diccionario.items()
print(equipo.items())
print(len(equipo)) # Obtenemos el número de elementos (clave:valor) que tenemos en el diccionario
# En caso de querer vaciar por completo el diccionario ocupamos el método diccionario.clear()
equipo.clear() # Aquí borramos todos los elementos del diccionario
print(equipo)
```

```python
'''
32. Programación en Python | Pilas (Stack)
'''
# Son una estructura de datos conocidas en muchos lenguajes de programación como 'Stack'
# En Python no existen directamente, las podemos simular haciendo el uso de listas
pila = [1,2,3]
print(pila)
# Agregamos elementos al final de la lista
pila.append(4)
pila.append(5)
print(pila)
# Sacamos elementos del final de la lista
n = pila.pop()
print(f'Sacando el elemento {n}')
print(pila)
```

```python
'''
33. Programación en Python | Colas
'''
# Son una estructura de datos tipo 'fifo' (first input, first output)
# Vamos a simular una cola en espera en el banco
cola = ['Eduardo','Kenya','José','Paola']
print(cola)
# Si llegan más personas se agregan al final de la cola
cola.append('Karla')
cola.append('David')
print(cola)
# Sacando elementos por el principio de la cola
n = cola.pop(0) # Sacamos el elemento con el primer índice de la lista
print(f'Atendiendo a {n}') # Nos muestra el elemento que atendieron y sale de la cola del banco
print(cola) # Nos muestra la cola sin el primer elemento que ya fue atendido
n = cola.pop(0) # Sacamos el elemento con el primer índice de la lista
print(f'Atendiendo a {n}') # Nos muestra el elemento que atendieron y sale de la cola del banco
print(cola) # Nos muestra la cola sin el primer elemento que ya fue atendido
```

```python
'''
34. Programación en Python | Ejercicio 1
'''
# Escriba un programa donde tenga una lista y que borre los elementos repetidos de la lista
lista = [1,2,3,4,5,'Eduardo',1,3,4,'Eduardo']
print(lista)
# Ocuparemos los conjuntos debido a que NO permiten los elementos duplicados
conjunto = set(lista) # Transformamos la lista en un conjunto con la nueva variable
print(conjunto) # Obtenemos un conjunto que no tiene elementos repetidos
lista = list(conjunto) # Transformamos la variable conjunto nuevamente en una lista
print(lista) # Obtenemos el resultado deseado con una lista sin elementos repetidos
# Si nos ponemos más creativos podemos transformar todas esas líneas de código en una sola línea
# Volvemos a crear la lista original de la cual eliminaremos sus elementos repetidos
lista = [1,2,3,4,5,'Eduardo',1,3,4,'Eduardo']
print(lista)
# Vamos a sobreescribir la lista para transformarla en un conjunto y ese conjunto nuevamente en una lista
lista = list(set(lista)) # En una sola línea de código hemos logrado eliminar los elementos repetidos
print(lista)
```

```python
'''
35. Programación en Python | Ejercicio 2
'''
# Escribe un programa que tenga 2 listas y cree las siguientes listas sin repeticiones:
# 1.Lista de elementos que aparecen en las dos listas
# 2.Lista de elementos que aparecen en la primera lista pero no en la segunda
# 3.Lista de elementos que aparecen en la segunda lista pero no en la primera
# 4.Lista de elementos que aparecen en ambas listas a la vez
lista1 = [1,2,3,4,5,4,3,2,2,1,5]
lista2 = [4,5,6,7,8,4,5,6,7,7,8]
# Primero vamos a eliminar los elementos repetidos
a = set(lista1)
b = set(lista2)

union = list(a|b) #A∪B
soloA = list(a-b) # A\B
soloB = list(b-a) # B\A
interseccion = list(a&b) # A∩B

print(f'Lista de elementos que aparecen en las dos listas: {union}') #A∪B
print(f'Lista de elementos que aparecen en la primera lista pero no en la segunda: {soloA}') # A\B
print(f'Lista de elementos que aparecen en la segunda lista pero no en la primera: {soloB}') # B\A
print(f'Lista de elementos que aparecen en ambas listas a la vez: {interseccion}') # A∩B
```
