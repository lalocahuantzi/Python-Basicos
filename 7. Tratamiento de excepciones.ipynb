#### Errores Sintácticos:
Los errores de sintaxis ocurren cuando el código no respeta la grámatica y la estructura de un lenguaje, como un parentésis sin cerrar o un punto y coma ausente, impidiendo su compilación o interpretación.

#### Errores Semánticos

Los errores de semántica se producen cuando el código es sintácticamente correcto pero carece de sentido o lógica, resultando en un comportamiento indesperado o un resultado incorrecto, como intentar sumar un número y una cadena de texto.

```python
'''
72. Programación en Python | Excepciones
'''

while True: # Hasta que la suma se ejecute de manera correcta terminará, si procede la excepción entonces el programa sigue
    try: # Instrucciones a intentar realizar
        numero = int(input("Digite un número: ")) # Si el usuario ingresa un valor o caracter incorrecto esta línea falla
        print(f'La suma es {numero+10}')
        break # Si el 'try' es correcto entonces el bucle While se rompe y el programa finaliza
    except: # Si las instrucciones del 'try' fallan, el programa procede con la excepción
        print('Ha ocurrido un error')

print("Programa terminado") # Cuando se rompe el bucle anunciamos que el programa finalizó
```

```python
'''
73. Programación en Python | Ejemplo con Múltiples Excepciones
'''
# Tenemos que ser más específicos con el except para que el usuario conozca lo que se está haciendo mal
def dividir():
    while True:
        try:
            num1 = float(input('Digite un número: '))
            num2 = float(input('Digite otro número: '))
            resultado = num1/num2
            print(f'El resultado de la división es: {resultado:.3f}')
        except ValueError: # En caso de que el usuario ingrese un caracter incorrecto se ejecuta esta excepción
            print('Error -> Por favor digite correctamente los valores numéricos')
        except ZeroDivisionError: # En caso de que el usuario ingrese un 0 en el segundo valor se ejecuta esta excepción
            print('Error -> No se puede dividir entre 0')
        else:
            break

dividir() # Llamamos a la función para que se ejecute
```

```python
'''
74. Programación en Python | Lanzar nuestras propias excepciones
'''
# Podemos lanzar nuestras propias excepciones con el comando 'raise'
def verificandoEdad(edad):
    if edad < 0: # En caso de que la edad sea negativa
        raise ValueError('La edad NO puede ser negativa') # Lanzamos nuestra propia excepción con una edad negativa
    elif edad < 18:
        print('Eres muy joven')
    elif edad <30:
        print('Eres joven')
    elif edad <50:
        print('Eres maduro')

edad = int(input('Digite su edad: '))
try:
    verificandoEdad(edad) 
except ValueError as EdadNegativa: # Podemos ocupar un alias para la misma excepción que nosotros habíamos lanzando
    print(EdadNegativa) # Imprimimos en pantalla la excepción en caso de que se dígite una edad negativa

print('Programa terminado')
```
